src/

<div style="display:flex;flex-wrap:wrap;gap:10px;margin:6px 0 14px 0;font-family:-apple-system,Segoe UI,Roboto,sans-serif"><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">O(y)</div><div style="font-size:12px;color:#52514e;margin-top:3px">tempo e memória da pilha</div></div><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">y + 1</div><div style="font-size:12px;color:#52514e;margin-top:3px">chamadas para mult(x, y)</div></div><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">no retorno</div><div style="font-size:12px;color:#52514e;margin-top:3px">momento em que a soma acontece</div></div><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">961 / 961</div><div style="font-size:12px;color:#52514e;margin-top:3px">pares conferidos com x * y</div></div></div>

#### 📐 Fundamentação

Uma função recursiva precisa de duas partes: o **caso base**, que responde sem chamar a si mesma, e o
**caso recursivo**, que reduz o problema em direção a ele (WENGROW, 2023, cap. 10). Aqui:

<table style="margin-left:0;margin-right:auto;border-collapse:collapse;font-size:13px"><thead><tr><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Parte</th><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Definição</th></tr></thead><tbody><tr><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">caso base</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df"><code>mult(x, 0) = 0</code></td></tr><tr><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">caso recursivo</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df"><code>mult(x, y) = x + mult(x, y − 1)</code>, para y > 0</td></tr></tbody></table>

> **Decisões.** y precisa ser inteiro e não negativo: com y negativo, `y − 1` nunca chega a 0 e a
> recursão não teria fim, então a função recusa com `ValueError`. x pode ser qualquer inteiro, porque
> só entra nas somas. O rastro é um callback (Observer): a função avisa cada chamada, caso base e
> retorno, e quem imprime é o script.

#### 📊 Análise

**Quando a soma acontece.** Em `x + mult(x, y − 1)` o Python precisa do valor da chamada interna
antes de somar. Por isso as 5 chamadas de `mult(5, 4)` descem sem fazer conta nenhuma, o caso base
devolve 0 e as 4 somas acontecem **no retorno**, de baixo para cima: 5 + 0, 5 + 5, 5 + 10, 5 + 15. Na
versão acumulada a soma vai no argumento, **na chamada**, e a volta só repassa o 20 pronto.

**Custo.** São y + 1 chamadas e y somas, e a pilha chega a y + 1 quadros: **O(y)** em tempo e em
memória. Como x · y = y · x, passar o menor dos dois como y reduziria a profundidade, mas o
enunciado fixa a ordem dos parâmetros.

As soluções dos dois ramos, juntas, são a resposta (CANNING; BRODER; LAFORE, 2022, cap. 6).

> conjunto de itens: `[5, 5, 10]` com alvo 10 tem duas soluções, `[5 5]` e `[10]`. Peso zero ou

## 8 · Exercício 7 · Análise de complexidade em algoritmos recursivos

<img src="https://img.shields.io/badge/tema-programa%C3%A7%C3%A3o%20din%C3%A2mica-EB6834?style=flat-square" /> <img src="https://img.shields.io/badge/estrutura-knapsack%20%28Ex.%206%29-52514E?style=flat-square" /> <img src="https://img.shields.io/badge/Big%20O-O%282%E2%81%BF%29-1E3A8A?style=flat-square" />

> A mochila resolve o mesmo subproblema milhares de vezes: 99,9% das chamadas são repetidas.

<div style="display:flex;flex-wrap:wrap;gap:10px;margin:6px 0 14px 0;font-family:-apple-system,Segoe UI,Roboto,sans-serif"><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">99,9%</div><div style="font-size:12px;color:#52514e;margin-top:3px">chamadas redundantes com n = 20</div></div><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">1.310</div><div style="font-size:12px;color:#52514e;margin-top:3px">estados distintos para 1.587.347 chamadas</div></div><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:16px;font-weight:700;color:#1E3A8A;line-height:1.25">T(n) = 2T(n−1) + 1</div><div style="font-size:12px;color:#52514e;margin-top:3px">recorrência</div></div><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">×1.024</div><div style="font-size:12px;color:#52514e;margin-top:3px">chamadas ao dobrar n de 10 para 20</div></div></div>

#### 📐 Chamadas redundantes e recorrência

Escolhi o `knapsack` do Exercício 6. O resultado de uma chamada depende só de dois valores: o **alvo
que falta** e o **índice** do próximo peso. Esse par é o **estado** do subproblema. Caminhos diferentes
chegam ao mesmo estado: em `knapsack(6, [1, 2, 3, 4])`, pegar 1 e 2 ou pegar só o 3 deixa os dois com
"falta 3, a partir do índice 3". A recursão pura não sabe disso e refaz a subárvore inteira em cada
visita (WENGROW, 2023, cap. 12).

**Recorrência.** No pior caso (nenhum corte), cada chamada faz duas sobre n − 1 pesos:

T(n) = 2 · T(n − 1) + 1, com T(0) = 1, logo T(n) = 2ⁿ⁺¹ − 1 e o custo é **O(2ⁿ)**.

Já os estados distintos são no máximo (n + 1) · (2 · alvo + 1), contando os alvos negativos dos
cortes: um número **polinomial** em n e no alvo. A diferença entre as duas contagens é trabalho
repetido.

#### 🧩 Implementação

O núcleo não muda: o `knapsack` aceita um callback `ao_visitar`, e o script conta as visitas a cada estado num `Counter`.

#### 📐 Fundamentação

A programação dinâmica se aplica quando o problema tem **subproblemas sobrepostos**, como mostrou o
Exercício 7, e a solução se monta a partir das soluções dos subproblemas (BELLMAN, 1957; CORMEN et
al., 2012, cap. 15). A **memoização** mantém a recursão de cima para baixo e acrescenta um dicionário:
antes de resolver um estado, a função consulta o dicionário; depois de resolver, guarda a resposta
(WENGROW, 2023, cap. 12).

<table style="margin-left:0;margin-right:auto;border-collapse:collapse;font-size:13px"><thead><tr><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Versão</th><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Chamadas</th><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Big O</th></tr></thead><tbody><tr><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">recursão pura</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">até 2ⁿ⁺¹ − 1</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">O(2ⁿ)</td></tr><tr><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">com memo</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">no máximo 2 por estado guardado, mais a inicial</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">O(n · alvo)</td></tr></tbody></table>

> **Decisões.** Guardo a lista de combinações de cada estado, então a versão memo devolve exatamente
> as mesmas soluções da recursão pura. Quando há muitas soluções, montar as listas também custa; por
> isso incluí `contar_combinacoes`, que guarda só a quantidade e isola o custo dos estados. O(n · alvo)
> é **pseudopolinomial**: polinomial no valor do alvo, não no número de dígitos dele.

#### 🧩 Implementação

## 10 · Exercício 9 · QuickSort com o último elemento como pivô

<img src="https://img.shields.io/badge/tema-ordena%C3%A7%C3%A3o%20e%20sele%C3%A7%C3%A3o-6250D6?style=flat-square" /> <img src="https://img.shields.io/badge/estrutura-array-52514E?style=flat-square" /> <img src="https://img.shields.io/badge/Big%20O-O%28N%20log%20N%29%20a%20O%28N%C2%B2%29-1E3A8A?style=flat-square" />

#### 📐 Fundamentação

A **partição** escolhe o último elemento como pivô e usa dois ponteiros: o da esquerda avança
enquanto encontra menores que o pivô, o da direita recua enquanto encontra maiores. Quando os dois
param, os valores trocam de lado; quando se cruzam, o pivô vai para onde o da esquerda parou, que é a
sua posição final. O **QuickSort** particiona e chama a si mesmo para o trecho à esquerda e para o
trecho à direita do pivô; o caso base é um trecho com 0 ou 1 elemento (WENGROW, 2023, cap. 13;
HOARE, 1962).

<table style="margin-left:0;margin-right:auto;border-collapse:collapse;font-size:13px"><thead><tr><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Caso</th><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Quando acontece</th><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Profundidade</th><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Big O</th></tr></thead><tbody><tr><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">melhor e médio</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">o pivô cai perto do meio</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">≈ log₂ N</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">O(N log N)</td></tr><tr><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">pior</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">o pivô é sempre o menor ou o maior</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">N</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">O(N²)</td></tr></tbody></table>

> **Decisões.** Os ponteiros param também em valores **iguais** ao pivô. Isso gera trocas a mais com
> repetidos, mas divide ao meio uma lista de valores iguais, em vez de cair no pior caso. A ordenação
> acontece numa cópia, então a lista de entrada não muda.

## 12 · Exercício 11 · QuickSelect

<img src="https://img.shields.io/badge/tema-ordena%C3%A7%C3%A3o%20e%20sele%C3%A7%C3%A3o-6250D6?style=flat-square" /> <img src="https://img.shields.io/badge/estrutura-parti%C3%A7%C3%A3o-52514E?style=flat-square" /> <img src="https://img.shields.io/badge/Big%20O-O%28N%29%20m%C3%A9dio-1E3A8A?style=flat-square" />

> A mesma partição do QuickSort, mas a recursão segue só pelo lado que contém o índice procurado.

<div style="display:flex;flex-wrap:wrap;gap:10px;margin:6px 0 14px 0;font-family:-apple-system,Segoe UI,Roboto,sans-serif"><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">O(N)</div><div style="font-size:12px;color:#52514e;margin-top:3px">caso médio</div></div><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">2,7 · N</div><div style="font-size:12px;color:#52514e;margin-top:3px">comparações para a mediana de 100.001</div></div><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">6.460 / 6.460</div><div style="font-size:12px;color:#52514e;margin-top:3px">índices iguais ao sorted()[k]</div></div><div style="flex:1 1 170px;padding:10px 14px;border-radius:10px;background:#f6f8fc;border-left:4px solid #1E3A8A"><div style="font-size:19px;font-weight:700;color:#1E3A8A;line-height:1.25">para no pivô</div><div style="font-size:12px;color:#52514e;margin-top:3px">quando a posição do pivô é k</div></div></div>

#### 📐 Fundamentação

Depois de uma partição, o pivô está na posição definitiva: à esquerda ficam os menores ou iguais e à
direita os maiores ou iguais. Para achar o valor do índice k da lista ordenada (o k-ésimo menor),
basta comparar k com a posição do pivô (HOARE, 1961; WENGROW, 2023, cap. 13):

<table style="margin-left:0;margin-right:auto;border-collapse:collapse;font-size:13px"><thead><tr><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Situação</th><th style="text-align:left;padding:7px 12px;background:#1E3A8A;color:#fff">Ação</th></tr></thead><tbody><tr><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">posição do pivô = k</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df"><b>para</b>: o pivô é a resposta</td></tr><tr><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">k < posição do pivô</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">continua só à esquerda</td></tr><tr><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">k > posição do pivô</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">continua só à direita</td></tr><tr><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">trecho com 1 elemento</td><td style="text-align:left;padding:6px 12px;border-bottom:1px solid #e4e3df">caso base: é a resposta</td></tr></tbody></table>

Cada partição descarta um lado inteiro. No caso médio, o trabalho é N + N/2 + N/4 + … ≈ 2N: **O(N)**.
No pior caso, o mesmo do QuickSort, O(N²).

> **Decisões.** k é o índice na lista ordenada, de 0 a N − 1; fora disso, `IndexError`. A mediana usa
> k = N // 2 para N ímpar e a média dos dois centrais para N par.

````python
codigo(quickselect, mediana)
````

````python
tamanhos = [1_001, 10_001, 100_001]
ordenacao, selecao = [], []
for n in tamanhos:
    _, (comp_ordenacao, comp_selecao, _, _) = ex12.medias(n)
    ordenacao.append(comp_ordenacao); selecao.append(comp_selecao)
fig, ax = plt.subplots(figsize=(8.6, 3.6))
x = range(len(tamanhos))
for desloc, valores, cor, nome in ((-0.2, ordenacao, LARANJA, 'QuickSort (ordena tudo)'), (0.2, selecao, AZUL, 'QuickSelect (só a mediana)')):
    barras = ax.bar([i + desloc for i in x], valores, 0.38, color=cor, label=nome)
    for b, v in zip(barras, valores):
        ax.text(b.get_x() + b.get_width() / 2, v * 1.15, milhar(v), ha='center', fontsize=8, color=TINTA_2)
ax.set_yscale('log'); ax.set_ylim(1_000, 1e7)
ax.set_xticks(list(x), [f'N = {milhar(n)}' for n in tamanhos])
ax.set_title('Comparações para achar a mediana (média de 10 listas)'); ax.set_ylabel('comparações (escala log)'); ax.legend(loc='upper left')
mostrar('ex12_quicksort_x_quickselect')
legenda_grafico('Gráfico 5. A distância entre as barras aumenta com N: é o fator log N que o QuickSelect não paga.')
````

#### 📊 Análise

**Comparações e cópias.** O QuickSelect fez de **3,6 a 6,9 vezes menos comparações** e de 3,9 a 7,1
vezes menos cópias. A razão cresce com N, como esperado de N log N ÷ N = log N. Com 100.001 elementos,
até 6 seleções avulsas ainda custam menos que uma ordenação completa.

**Pior caso.** Na lista ordenada os dois degradam: o QuickSort faz ≈ N²/2 comparações e o QuickSelect
≈ 3N²/8, porque o pivô só alcança o meio depois de N/2 partições que descartam um elemento cada.

## 16 · Conclusão

1. Toda recursão deste TP começa pelo caso base e por uma regra que garante chegar até ele: y diminui,
   o quociente diminui, o expoente cai pela metade, o diretório não tem subdiretórios, os candidatos ou
   os pesos acabam, o trecho fica com um elemento.
2. A forma da recursão decide a complexidade. Uma chamada que reduz em 1 é linear, uma que divide por
   2 é logarítmica, e duas que reduzem em 1 dobram a árvore a cada nível. A mochila mostrou o preço: 2ⁿ
   chamadas para poucos milhares de estados distintos.
3. Memoização troca memória por tempo quando os subproblemas se repetem. Guardando cada estado uma vez,
   a mochila passou de O(2ⁿ) para O(n · alvo) sem mudar o resultado.
4. QuickSort e QuickSelect usam a mesma partição e diferem numa decisão: seguir pelos dois lados ou por
   um. Ordenar custa O(N log N); selecionar, O(N). Com o último elemento como pivô, os dois ficam
   quadráticos em entradas ordenadas, o que as contagens de comparações deixaram explícito.
